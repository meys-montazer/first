```python
price = float(input('enter the price of your shoping: '))

if price >= 100:
    off = price - (0.20 * price)
    print(f'you have to pay {off}')

elif price < 100 and price >= 50:
    off = price - (0.10 * price)
    print(f'you have to pay {off}')

else:
    print(f'you have to pay {price}')
```
